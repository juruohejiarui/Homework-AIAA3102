## Task 1: Sigmoid probabilities
Given logits, use `map` + `lambda` to compute $\sigma(x) = 1 / (1 + e^{-x})$ for each.


```python
import math

logits = [2.0, -1.0, 0.5, -2.5, 3.0]

# Using map + lambda
probabilities = list(map(lambda x: 1 / (1 + math.exp(-x)), logits))

print("Logits:", logits)
print("Probabilities (Sigmoid):", [round(p, 4) for p in probabilities])
```

## Task 2: Filter predictions
From a list of dicts with `{"label": str, "conf": float}`, keep only items with `conf >= 0.9` using `filter` + `lambda`.


```python
predictions = [
    {"label": "cat", "conf": 0.95},
    {"label": "dog", "conf": 0.85},
    {"label": "bird", "conf": 0.92},
    {"label": "horse", "conf": 0.70},
    {"label": "car", "conf": 0.99}
]

# Using filter + lambda
high_conf_preds = list(filter(lambda x: x["conf"] >= 0.9, predictions))

print("Filtered Predictions (conf >= 0.9):")
for p in high_conf_preds:
    print(p)
```

## Task 3: Rank object detection results
Use `sorted` with a `lambda` key to sort detections by confidence (descending). Break ties by larger box area.


```python
detections = [
    {"id": 1, "conf": 0.9, "area": 150},
    {"id": 2, "conf": 0.8, "area": 300},
    {"id": 3, "conf": 0.9, "area": 200},
    {"id": 4, "conf": 0.95, "area": 100},
    {"id": 5, "conf": 0.8, "area": 250}
]

# Using sorted + lambda
# To sort by conf descending, then area descending, we can return a tuple with negative values/reverse=True
ranked_detections = sorted(detections, key=lambda x: (x["conf"], x["area"]), reverse=True)

print("Ranked Detections:")
for d in ranked_detections:
    print(d)
```

## Task 4: Rewrite and Compare
Rewrite one subtask using a list comprehension or a normal def function. Which version is clearer and why?


```python
# Rewriting Task 2 (Filter predictions) using a list comprehension

# Original (filter + lambda):
# high_conf_preds = list(filter(lambda x: x["conf"] >= 0.9, predictions))

# Rewritten (List Comprehension):
high_conf_preds_comp = [x for x in predictions if x["conf"] >= 0.9]

print("Filtered Predictions (List Comprehension):")
for p in high_conf_preds_comp:
    print(p)
```

### Comparison Answer:

I chose to rewrite the **filtering task** using a **list comprehension**.

**Which version is clearer?**
The **list comprehension** (`[x for x in predictions if x["conf"] >= 0.9]`) is clearer and more readable.

**Why?**
1. **Pythonic Syntax**: It reads almost like plain English ("Give me x for every x in predictions if its confidence is >= 0.9").
2. **No extra function calls**: It avoids the need to wrap the result in `list()` which is required when using `filter()` in Python 3.
3. **Less boilerplate**: It avoids the `lambda` keyword entirely, making the expression more concise and intuitive.
